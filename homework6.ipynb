W6A16 Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma ̣
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n, m = map(int, input("Nhập n và m: ").split())
matrix = []
for i in range(n):
    row = list(map(int, input(f"Nhập hàng {i+1}: ").split()))
    matrix.append(row)
print("\nKết quả ma trận:")
for row in matrix:
    for val in row:
        print(f"{val:>4}", end="")
    print() 
```

Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng

```python
n = int(input())
matrix = [list(map(int, input().split())) for _ in range(n)]
chinh = [matrix[i][i] for i in range(n)]
phu = [matrix[i][n - 1 - i] for i in range(n)]
print(*chinh)
print(*phu)
```

Nhâ ̣p vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên ̣
cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột
k (hàng k)

```python
n, m, k = map(int, input().split())
matrix = [list(map(int, input().split())) for _ in range(n)]
tong_hang_k = sum(matrix[k])
tong_cot_k = sum(matrix[i][k] for i in range(n))

print(f"Tổng hàng {k}: {tong_hang_k}")
print(f"Tổng cột {k}: {tong_cot_k}")
```

W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối



```python
#Bài W6A26
a = list(map(int, input().split()))
n = len(a)
tong_cuc_dai = 0
for i in range(1, n - 1):
    if a[i] > a[i-1] and a[i] > a[i+1]:
        tong_cuc_dai += a[i]

print(tong_cuc_dai)
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm)


```python
def check_monotonicity(a):
    n = len(a)
    if n < 2:
        return "Tang thuc su" 

    is_increasing = True
    is_decreasing = True

    for i in range(n - 1):
        if a[i] >= a[i+1]: 
            is_increasing = False
        if a[i] <= a[i+1]: 
            is_decreasing = False

    if is_increasing:
        return "Tang thuc su"
    elif is_decreasing:
        return "Giam thuc su"
    else:
        return "Vo danh"

# Nhập dữ liệu
a = list(map(int, input().split()))
print(check_monotonicity(a))
    
        
```
