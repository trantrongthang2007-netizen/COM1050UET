Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
a = list(map(int, input().split()))
# a = [1, 2, 3, 4, 4, 1, 2, 3]
cnt = {}

for x in a:
    if x in cnt:
        cnt[x] += 1
    else:
        cnt[x] = 1

y = 0
# print(cnt)
# print(list(cnt.keys()))
# print(list(cnt.values()))

for i,x in cnt.items():
    if i < x:
        y += x - i
    elif i > x:
        y += x
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
n = int(input())
a = list(map(int, input().split()))

cnt = 0
candidate = -1

for x in a:
    if cnt == 0:
        candidate = x
        cnt = 1
    elif x == candidate:
        cnt += 1
    else:
        cnt -= 1

if a.count(candidate) > n // 2:
    print(candidate)
else:
    print(-1)
```

W7A18 Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần;

W7A19
Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không dùng bất cứ thuật toán sắp xếp nào


```python
n = int(input())
a = list(map(int, input().split()))

m = int(input())
b = list(map(int, input().split()))

tang = True

if n > 1:
    tang = (a[0] <= a[-1])
elif m > 1:
    tang = (b[0] <= b[-1])

i = 0
j = 0
ans = []

while i < n and j < m:
    if tang:
        if a[i] <= b[j]:
            ans.append(a[i])
            i += 1
        else:
            ans.append(b[j])
            j += 1

    else:
        if a[i] >= b[j]:
            ans.append(a[i])
            i += 1
        else:
            ans.append(b[j])
            j += 1

while i < n:
    ans.append(a[i])
    i += 1

while j < m:
    ans.append(b[j])
    j += 1

print(*ans)
```

Cho dãy N số nguyên dương, tính tổng các số có k chữ số. Dòng đầu ghi N số nguyên dương, dòng sau ghi số nguyên dương k


```python
a = list(map(int,input().split()))
k = int(input())

def check(x):
    tong = 0
    while x > 0:
        tong += 1
        x //= 10
    if tong == k:
        return True
    else:
        return False
res = 0
for i in a:
    if check(i):
        res += i
print(res)
```

Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
N = int(input())
name = []

for i in range(N):
    s = input()
    name.append(s)

cnt = {}
ten = []

for s in name:
    ten.append(s.split()[-1])

ten.sort()

for s in ten:
    if s in cnt:
        cnt[s] += 1
    else:
        cnt[s] = 1

Max = -10000005
ans = 0
print(cnt)
for id, val in cnt.items():
    if Max < val:
        Max = val
        ans = id
print(ans)
```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ hay không


```python
s = input()
stack = []

for x in s:
    if x in ['(','{','[']:
        stack.append(x)
    elif x in [')','}',']']:
        if len(stack) == 0:
            print("khong hop le")
            break
        if (x == ')' and stack[-1] == '(') or (x == ']' and stack[-1] == '[') or (x == '}' and stack[-1] == '{'):
            stack.pop()
else:
    if len(stack) == 0:
        print("hop le")
    else:
        print('khong hop le')
```

Bai toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ 
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
def check(n):
    if n % 2 == 0 and n != 2:
        return False
    if n < 2:
        return False
    for i in range(2,int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

a = list(map(int,input().split()))
n = int(input())

# a = [2,3,5,6,7,9,9,10,11,12,13,14,15,15,15,15]
# n = 5
# Output 14 15 15 15

id = []
for i in range(len(a)):
    if a[i] > n and check(a[i]) == False:
        id.append(i)
start = 0
cnt , Max = 1,1

id.append(None)
for i in range(1,len(id)):
    if id[i] == id[i-1] + 1:
        cnt += 1
    else:
        if cnt > Max:
            Max = cnt
            start = i - cnt - 1
        cnt = 1

print(a[id[start]:id[start + Max ]])
```

Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ 
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau


```python
a = list(map(int,input().split()))

cnt, Max = 1, 1
start = 0
# 3 3 3 2 3 4 5 6 7 9 4
for i in range(1,len(a)):
    if a[i] > a[i-1]:
        cnt += 1
    else:
        if cnt > Max:
            Max = cnt
            start = i - cnt
        cnt = 1
print(a[start:start + Max])
```

W7A30
Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 

```python
a = list(map(int,input().split()))
n = len(a)
dp = []
trace = []
for i in range(n):
    dp.append(1)
    trace.append(-1)
for i in range(n):
    for j in range(i):
        if a[i] > a[j]:
            if dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1
                trace[i] = j

id = dp.index(max(dp))
ans = []
while id != -1:
    ans.append(a[id])
    id = trace[id]

# print(dp)
# print(trace)
ans.reverse()
print(ans)
```
