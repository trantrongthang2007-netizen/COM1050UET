```python
class Shape():
    def area(self):
        pass

class Rectangle(Shape):
    def __innit__(self,dai,rong):
        self.dai = dai
        self.rong = rong
    def area(self):
        return (self.dai + self.rong) * 2
    
class Circle(Shape):
    def __init__(self,ban_kinh):
        self.ban_kinh = ban_kinh
    def area(self):
        return 3.14 * self.ban_kinh**2

Type = input()
Do_dai = list(map(int,input().split()))

if Type == 'CIRCLE':
    shape = Circle(Do_dai[0])
else:
    shape = Rectangle(Do_dai[0],Do_dai[1])

print(f"{shape.area():.2f}")
```

```python
class Vehicle():
    def __init__(self,brand,speed):
        self.brand = brand 
        self.speed = speed
    def info(self):
        return (f"Brand: {self.brand}, Speed: {self.speed}")

class Car(Vehicle):
    def __init__(self,brand,speed,seats):
        super().__init__(brand,speed)
        self.seats = seats
    def info(self):
        return (f"Brand: {self.brand}, Speed: {self.speed}, Seats: {self.seats}")

s = input()
if s == "Car":
    infor = list(map(str,input().split()))
    output = Car(infor[0],infor[1],infor[2])
    print(output.info())
else:
    infor = list(map(str,input().split()))
    output = Vehicle(infor[0],infor[1])
    print(output.info())
```

```python
class Person():
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def Discribe(self):
        return f"Name: {self.name} Age: {self.age}"
    
class Student(Person):
    def __init__(self,name,age,school):
        super().__init__(name,age)
        self.school = school 
    def Discribe(self):
        return f"Name: {self.name} Age: {self.age} School: {self.school}"

s = input()
a = list(map(str,input().split()))
if s == "Person":
    info = Person(a[0],a[1])
    print(info.Discribe())
else:
    info = Student(a[0],a[1],a[2])
    print(info.Discribe())
```

```python
class Account():
    def __init__(self,balance):
        self.balance = balance
    def Deposit(self,x):
        self.balance += x
    def Get_balance(self):
        return self.balance
class Saving_account(Account):
    def __init__(self,balance,rate):
        super().__init__(balance)
        self.rate = rate
    def Add_interest(self):
        self.balance += self.balance * self.rate

balance, rate = map(float, input().strip().split())
acc = Saving_account(balance,rate)
acc.Add_interest()
print(f"{acc.Get_balance():.2f}")
```

```python
class Matrix2x2():
    def __init__(self,a,b,c,d):
        self.a = a
        self.b = b
        self.c = c
        self.d = d
    def trace(self):
        return self.a + self.d
    def det(self):
        return self.a*self.d - self.b*self.c

a, b, c, d = map(int,input().split())
alo = Matrix2x2(a,b,c,d)

print(alo.trace(),alo.det())
```

```python
class Product():
    def __init__(self,name,price):
        self.name = name
        self.price = price

class DiscountProduct(Product):
    def __init__(self,name,price,percent):
        super().__init__(name,price)
        self.percent = percent 
    def final_price(self):
        return f"{(int(self.price) * (100 - int(self.percent)) / 100):.2f}"

n,p,per = map(str,input().split())
alo = DiscountProduct(n,p,per)
print(alo.final_price())
```

```python
class Counter():
    def __init__(self,cnt):
        self.cnt = cnt
    def inc(self):
        self.cnt -= 1
    def dec(self):
        self.cnt += 1
    def value(self):
        return self.cnt
    
n = int(input())
a = list(map(str,input().split()))
alo = Counter(0)
for x in a:
    if x == "INC":
        alo.inc()
    else:
        alo.dec()

print(alo.value())
```

```python
class Time():
    def __init__(self,h,m,s):
        self.h = h
        self.m = m
        self.s = s
    def to_sec(self):
        return h*3600 + m*60 + s
    
h,m,s = map(int,input().split())
alo = Time(h,m,s)
print(alo.to_sec())
```

```python
class Stack():
    def __init__(self,stack):
        self.stack = []
    def push(self,x):
        self.stack.append(x)
    def pop(self):
        if self.stack:
            self.stack.pop()
    def top(self):
        return self.stack[-1]
    
n = int(input())
alo = Stack([])
for i in range(0,n):
    c = input().split()
    if c[0] == "PUSH":
        alo.push(c[-1])
    if c[0] == "POP":
        alo.pop()
    if c[0] == "TOP":
        print(alo.top())
```
